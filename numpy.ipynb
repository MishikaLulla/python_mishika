```python
import numpy as np
a=np.array([[1,2,3],[4,5,6]])
print(np.sum(a)) #sum of all the elements
```

```python
#mean = avg
a=np.array([[1,2,3],[4,5,6]])
print(np.mean(a))
```

```python
a=np.array([[1,2,3],[4,5,6]])
print(np.max(a))
print(np.min(a))
```

```python
#matrix op dot product
a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])
print(a.dot(b))
```

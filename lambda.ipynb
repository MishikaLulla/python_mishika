```python
#lambda function in python
#small functions ,short filter,used once,one argument
add=lambda x,y:x+y
print(add(4,6))
```

```python
multi=lambda x,y:x*y
print(multi(4,6))
```

```python
sq=lambda x:x**2
print(sq(6))
```

```python
#exception handling manage errors or unexpected situations
#try- may raise an exception
#except-how to handle error
#else-no exception
#finally-always run
```

```python
try:
    num=int(input('enter a num'))
    print(10/num)
except ZeroaDivisionError:
    print("cannot print")
else:
    print("success")
finally:
  print("done")
```

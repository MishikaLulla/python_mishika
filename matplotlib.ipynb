```python
import matplotlib.pyplot as plt
```

```python
#line plot
x=[1,2,3,4,5]
y=[10,12,2,4,14]
plt.plot(x,y)
plt.title("basic line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
#line plot
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.plot(x,y)
plt.title("basic line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
#line plot
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.plot(x,y,color='pink',linestyle='--',marker='o',linewidth=5)
plt.title("straight line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
#bar chart
categories=['A','B','C','D']
values=[10,20,30,40]
plt.bar(categories,values,color='pink')
```

```python
#bar chart
categories=['A','B','C','D']
values=[10,20,30,40]
plt.bar(categories,values,color='pink')
plt.title('bar chart')
plt.xlabel('category')
plt.ylabel('values')
```

```python
#scatter plot
x=[3,4,5,6,7]
y=[7,6,5,4,3]
plt.scatter(x,y,color='pink')
plt.title("scatter plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
```

```python
#list=mutuable 
print("list in python")
fruits=["litchi","mango","apple"]
fruits
fruits[0]
fruits[-1] #last element
fruits.append("orange") #add data
fruits
fruits.remove("orange") #remove data
fruits
```

```python
#read all elements using for loop
for fruit in fruits:
    print(fruit)
```

```python
#length of list
len(fruits)
```

```python
#existence of item in list using "in"
print("mango" in fruits)

```

```python
print("orange" in fruits)
```

```python
fruits.append("orange")
print("orange" in fruits)
```

```python
for x in fruits:
    print(x)
```

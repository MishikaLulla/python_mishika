```python

import numpy as np
a=np.array([[1,2,3],[4,5,6]])
print(a[1,2])
print(a[0,2])
```

```python
#slicing- a part
import numpy as np
a=np.array([[1,2,3],[4,5,6]])
print(a[1,:])
print(a[0,:])
```

```python
#array operations
a=np.array([1,2,3])
b=np.array([4,5,6])
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**2)
print(a**b)
print(b**3)
```

```python
a=np.array([1,2,3])
print(a>1)
print(a==1)
```

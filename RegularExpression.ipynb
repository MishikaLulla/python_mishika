```python
text="hi , i am mishika my age is 20 and number is 1234"
digits=re.findall(r"\d+",text) #find digits
digits
```

```python
#replacement
text="replace all spaces"
new_text=re.sub(r"\s","-",text)
new_text
```

```python
#replacement
text="replace all spaces"
new_text=re.sub(r"l","=",text)
new_text
```

```python
text="contact me at mishi@email.com"
email=re.findall(r"[a=zA-Z0-9._]+@[a=zA-Z0-9._]+\.[a-zA-Z]{2,}",text)
print(email)
```
